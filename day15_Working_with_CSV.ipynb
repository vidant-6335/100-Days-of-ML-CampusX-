## Opening a CSV file from a url

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
headers = {"User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10.14; rv:66.0) Gecko/20100101 Firefox/66.0"}
req = requests.get(url, headers=headers)
data = StringIO(req.text)

pd.read_csv(data)
```

## Sep Parameter

```python
pd.read_csv('movie_titles_metadata.tsv', sep='\t', names=['sno', 'name', 'release_year', 'rating', 'votes', 'genres'])
```

## Index_col parameter

```python
pd.read_csv('ecommerce.csv', index_col='order_id')
```

## Header parameter

```python
pd.read_csv('ecommerce.csv', header=1)
```

## use_cols parameter

```python
pd.read_csv('ecommerce.csv',usecols=['order_id', 'price', 'discount'])
```

## Squeeze parameters

```python
pd.read_csv('ecommerce.csv')['order_id']
```

## Skiprows/nrows Parameter

```python
pd.read_csv('ecommerce.csv', skiprows=[1, 3, 5])
```

## Encoding parameter

```python
pd.read_csv('zomato.csv', encoding='latin-1').head()
```

## Skip bad lines

```python
df = pd.read_csv('zomato.csv',sep=';',encoding='latin-1',on_bad_lines='skip',engine='python')
df
```

## dtypes parameter

```python
pd.read_csv('ecommerce.csv', dtype={'quantity':float}).info()
```

## Handling Dates

```python
# pd.read_csv by default read dates as strings
pd.read_csv('ecommerce.csv', parse_dates=['order_date']).info()
```

## Convertors

```python
def rename(name):
    if name == 'Home':
        return 'H'
    else:
        return name

pd.read_csv('ecommerce.csv', converters={'category':rename})
```

## na_values parameter

```python
#missing values
pd.read_csv('ecommerce.csv', na_values=['Male'])
#considering 'Male' as empty values 
```

## Loading a huge dataset in chunks

```python
dfs = pd.read_csv('ecommerce.csv', chunksize=5000)
for chunks in dfs:
    print(chunks.shape)
```
